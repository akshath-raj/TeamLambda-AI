```python
# Remove Unnamed columns
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

# Remove columns that are fully empty
df = df.dropna(axis=1, how="all")
```

```python
df = df.drop(columns=["Scan"], errors="ignore")
df.head()
```

```python
#unsoiled-0
#soiled-1
df.columns = [
    "Datetime",
    "V0",
    "V1",
    "C0",
    "C1"
]

df.head()
```

```python
num_cols = [
    "V0",
    "V1",
    "C0",
    "C1"
]

df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
```

```python
len(df)
```

```python
from datetime import datetime
```

```python
# Force Datetime column to string before string operations
raw_dates = (
    df["Datetime"]
    .astype(str)              # <-- FIX
    .str.strip()
    .str.replace("/", "-", regex=False)
)

raw_dates.head()

```

```python
# Known valid experiment date range
START_DATE = pd.Timestamp("2023-03-10 00:00:00")
END_DATE   = pd.Timestamp("2023-06-20 23:59:59")


START_DATE, END_DATE
```

```python
len(df)
```

```python
import warnings

parsed_dates = []
prev_time = START_DATE - pd.Timedelta(seconds=1)

with warnings.catch_warnings():
    warnings.simplefilter("ignore", UserWarning)

    for x in raw_dates:
        x = x.strip()

        # Try both interpretations
        dt_df = pd.to_datetime(x, dayfirst=True, errors="coerce")
        dt_mf = pd.to_datetime(x, dayfirst=False, errors="coerce")

        candidates = []

        for d in [dt_df, dt_mf]:
            if pd.notna(d):
                if START_DATE <= d <= END_DATE:
                    if d >= prev_time:
                        candidates.append(d)

        if candidates:
            chosen = min(candidates)
        else:
            # Hard safety to enforce monotonicity
            chosen = prev_time + pd.Timedelta(seconds=1)

        parsed_dates.append(chosen)
        prev_time = chosen
```

```python
len(df)
```

```python
df["Datetime"] = parsed_dates

df["Datetime"].is_monotonic_increasing
```

```python
len(df)
```

```python
df["Datetime"].min(), df["Datetime"].max()
```

```python
df = df.set_index("Datetime")

df.head()
```

```python
# Floor timestamps to hour (keeps only existing hours)
df["hour"] = df.index.floor("h")

# Aggregate ONLY existing hours
hourly_df = df.groupby("hour").mean(numeric_only=True)

hourly_df.head()
```

```python
len(hourly_df)
```

```python
hourly_df.head()
```

```python
#Prithika's part
```

```python
df2 = hourly_df.reset_index()
```

```python
unsoiled = df2[["hour", "V0", "C0"]].copy()
unsoiled.columns = ["Datetime", "Voltage", "Current"]
unsoiled["Panel_State"] = 0   # unsoiled
```

```python
soiled = df2[["hour", "V1", "C1"]].copy()
soiled.columns = ["Datetime", "Voltage", "Current"]
soiled["Panel_State"] = 1   # soiled
```

```python
unsoiled = unsoiled.reset_index(drop=True)
soiled   = soiled.reset_index(drop=True)
```

```python
rows = []
min_len = min(len(soiled), len(unsoiled))

for i in range(min_len):
    rows.append(soiled.iloc[i])
    rows.append(unsoiled.iloc[i])
```

```python
final_df = pd.DataFrame(rows)
```

```python
final_df["Power"] = final_df["Voltage"] * final_df["Current"]
```

```python
final_df = final_df[["Datetime", "Voltage", "Current", "Power", "Panel_State"]]
```

```python
final_df = final_df.set_index("Datetime")
```

```python
final_df = final_df.sort_index()
```

```python
final_df.reset_index().to_csv(
    "cleaned_panel_data_ps.csv",
    index=False,
)

print("✅ Cleaned hourly dataset saved")
```
